#✅ TRANSFORMATIONS
Transformations are lazy operations that define a new RDD/DataFrame from an existing one. Nothing is executed until an action is triggered.
RDD transforamtions




| Category | Transformation     | Description                                     |
| -------- | ------------------ | ----------------------------------------------- |
| Narrow   | `map()`            | Applies a function to each element              |
| Narrow   | `filter()`         | Filters elements based on condition             |
| Narrow   | `flatMap()`        | Similar to `map`, but flattens the result       |
| Narrow   | `mapPartitions()`  | Applies function to each partition              |
| Narrow   | `sample()`         | Randomly samples a subset                       |
| Narrow   | `union()`          | Combines two RDDs                               |
| Narrow   | `distinct()`       | Removes duplicates                              |
| Wide     | `groupByKey()`     | Groups values by key                            |
| Wide     | `reduceByKey()`    | Combines values by key using reduce             |
| Wide     | `aggregateByKey()` | Aggregate values by key with a custom function  |
| Wide     | `sortByKey()`      | Sorts by key                                    |
| Wide     | `join()`           | Joins two RDDs by key                           |
| Wide     | `cogroup()`        | Groups data from two RDDs sharing the same key  |
| Wide     | `cartesian()`      | Cartesian product                               |
| Narrow   | `pipe()`           | Pipe each partition through an external command |
| Narrow   | `repartition()`    | Increases/decreases partitions                  |
| Narrow   | `coalesce()`       | Reduces number of partitions                    |


# DataFrame / Dataset Transformations:

| Category    | Transformation                 | Description                         |
| ----------- | ------------------------------ | ----------------------------------- |
| Projection  | `select()`                     | Select specific columns             |
| Filtering   | `filter()` / `where()`         | Filter rows based on condition      |
| Mapping     | `withColumn()`                 | Adds/replaces a column              |
| Mapping     | `drop()`                       | Removes a column                    |
| Mapping     | `alias()`                      | Renames a column                    |
| Aggregation | `groupBy()`                    | Groups rows by key                  |
| Aggregation | `agg()`                        | Aggregate with functions            |
| Join        | `join()`                       | Joins DataFrames                    |
| Set Ops     | `union()`                      | Combines two DataFrames             |
| Set Ops     | `intersect()`                  | Intersect of two DataFrames         |
| Set Ops     | `except()`                     | Subtract one DataFrame from another |
| Sorting     | `orderBy()` / `sort()`         | Sorts the data                      |
| Window      | `over()`                       | Window functions                    |
| Misc        | `distinct()`                   | Removes duplicates                  |
| Misc        | `dropDuplicates()`             | Removes duplicate rows              |
| Misc        | `repartition()` / `coalesce()` | Changes partitioning                |
| Misc        | `limit()`                      | Limits number of rows               |
| Misc        | `na.drop()`, `na.fill()`       | Handles missing data                |


#✅ ACTIONS
Actions trigger the execution of transformations and return results.

| Action                 | Description                                             |
| ---------------------- | ------------------------------------------------------- |
| `collect()`            | Returns all elements as a list                          |
| `count()`              | Number of elements                                      |
| `take(n)`              | First `n` elements                                      |
| `top(n)`               | Top `n` elements                                        |
| `takeOrdered(n)`       | First `n` ordered elements                              |
| `reduce()`             | Reduces elements using a function                       |
| `fold()`               | Aggregates using an initial zero value                  |
| `aggregate()`          | Combines elements with two functions                    |
| `foreach()`            | Applies a function to each element (no result returned) |
| `saveAsTextFile()`     | Saves RDD to text files                                 |
| `saveAsSequenceFile()` | Saves RDD to Hadoop sequence file                       |
| `countByValue()`       | Count of each unique value                              |
| `lookup()`             | Lookup values for a key (for PairRDDs)                  |


| Action               | Description                                     |
| -------------------- | ----------------------------------------------- |
| `show()`             | Displays rows in tabular form                   |
| `collect()`          | Returns all rows as a list                      |
| `count()`            | Number of rows                                  |
| `first()` / `head()` | First row                                       |
| `take(n)`            | First `n` rows                                  |
| `foreach()`          | Applies function to each row                    |
| `foreachPartition()` | Applies function to each partition              |
| `write()`            | Writes to storage (CSV, Parquet, etc.)          |
| `toPandas()`         | Converts to Pandas DataFrame (beware of memory) |
| `summary()`          | Returns summary statistics                      |
| `describe()`         | Summary statistics for numeric columns          |
| `cache()`            | Caches the DataFrame                            |
| `persist()`          | Persists the DataFrame with storage level       |
