```python
df1 = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema",True)\
    .option("samplingRatio",0.10)\
    .load("/Volumes/datasets/practice/internal_files/orders.csv")
df1.count()
```

```python
df1.count()
```

#You get 68,884, as expected — because you haven’t sampled the data itself, only used a sampling ratio for schema inference.

#2 ways to Enforce the Schema

```python
orders_schema = 'Order_id long, order_date date, customer_id long, order_stats String'
```

```python
df4 = spark.read.format("csv")\
    .option("header", "true")\
    .schema(orders_schema)\
    .load("/Volumes/datasets/practice/internal_files/orders.csv")
df4.count()
df4.printSchema()
```

```python
df4.count()
```

```python
df4.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df4.withColumn("order_date", to_date(current_timestamp())).display()
```

```python
orders_schema_new = "order_id long, order_date date, customer_ids long, order_status long"
df5 = spark.read.format("csv")\
    .option("header", True)\
    .schema(orders_schema_new)\
    .load("/Volumes/datasets/practice/internal_files/orders.csv")
df5.count()
df5.printSchema()
```

```python
df5.show()
```

#Struct Type

```python
order_schema_struct = StructType([
    StructField("order_id", LongType()),
    StructField("order_date", DateType()),
    StructField("Customer_id", IntegerType()),
    StructField("Order_status", StringType())
])
```

```python
df6 = spark.read.format("csv")\
    .option("header", True)\
    .schema(order_schema_struct)\
    .load("/Volumes/datasets/practice/internal_files/orders.csv")
```

```python
df6.printSchema()
```

```python
df6.display()
```
